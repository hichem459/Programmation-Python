## **3.3 Entry (champ de saisie)**
### **2.3.1 Entry (sans StringVar)**


**1️⃣ get() – Récupérer une valeur**

La méthode **get()** permet de lire le contenu d’un widget.

**📌 Avec Entry**

````python
import tkinter as tk

fenetre = tk.Tk()

label = tk.Label(fenetre, text="User Name", font=("Arial", 16))
label.pack()

entree = tk.Entry(fenetre)
entree.pack()


def afficher_texte():
    print(entree.get())


btn = tk.Button(fenetre, text="Afficher", command=afficher_texte)
btn.pack()

fenetre.mainloop()
````

**📌 Avec Text**

````python
import tkinter as tk

fenetre = tk.Tk()


entree= tk.Text(fenetre)
entree.pack()

def afficher_texte():
    print(entree.get("1.0", tk.END))  # Récupère tout le texte du widget Text

btn = tk.Button(fenetre, text="Afficher", command=afficher_texte)
btn.pack()

fenetre.mainloop()
````

**📌 Avec Listbox**

````python
import tkinter as tk

fenetre = tk.Tk()

listbox = tk.Listbox(fenetre)
listbox.pack()

# ajouter quelques éléments pour pouvoir sélectionner quelque chose
for item in ["Python", "Java", "C++"]:
	listbox.insert(tk.END, item)

def afficher_selection():
	sel = listbox.curselection()
	if sel:                    # on a bien un indice
		print(listbox.get(sel[0]))
	else:
		print("aucune sélection")

btn = tk.Button(fenetre, text="Afficher", command=afficher_selection)
btn.pack()


fenetre.mainloop()
````

**2️⃣ insert() – Insérer du texte**

La méthode **insert(index, valeur)** permet d’insérer du texte dans un widget.

**📌 Avec Entry**

````python
import tkinter as tk

fenetre = tk.Tk()

label = tk.Label(fenetre, text="User Name", font=("Arial", 16))
label.pack()

entree = tk.Entry(fenetre)
entree.pack()

entree.insert(0, "User Name")   # insère au début
#entree.insert(tk.END, "User Name")  # insère à la fin


def afficher_texte():
    print(entree.get())


btn = tk.Button(fenetre, text="Afficher", command=afficher_texte)
btn.pack()

fenetre.mainloop()
````

**🔎 Explication :**

**0 → position de départ**

**tk.END → fin du texte**

**📌 Avec Text**

````python
import tkinter as tk

fenetre = tk.Tk()

entree= tk.Text(fenetre)
entree.pack()

entree.insert("1.0", "Première ligne\n")

def afficher_texte():
    print(entree.get(index1="1.0", index2=tk.END))


btn = tk.Button(fenetre, text="Afficher", command=afficher_texte)
btn.pack()


fenetre.mainloop()
````

**🔎 Explication :**

**"1.0" → ligne 1, caractère 0**

**"2.0" → ligne 2, caractère 0**

**📌 Avec Listbox**

````python
import tkinter as tk

fenetre = tk.Tk()

listbox = tk.Listbox(fenetre)
listbox.pack()

listbox.insert(tk.END, "Python")
listbox.insert(tk.END, "Java")


fenetre.mainloop()
````

 # **Liste des options pour le widget Entry**

1. `bg` : définit la couleur d’arrière-plan du widget Entry.
2. `bd` : définit la taille des bordures. La valeur par défaut est 2 pixels.
3. `command`: définit et associe une commande
4. `cursor` : définit le motif du curseur de la souris
5. `font` : définit la police qui sera utilisée pour le texte.
6. `exportselection` : par défaut, si vous sélectionnez du texte dans un widget Entry, il
peut être exporté vers le Presse-papiers. Pour éviter cette exportation, utilisez exportselection = 0.
7. `fg` : définit la couleur qui sera utilisée pour le texte
8. `highlightcolor` : Il représente la couleur de surbrillance à utiliser pour le rectangle qui
est dessiné autour du widget lorsqu’il a le focus.
9. `justify` : Il spécifie comment le texte est organisé s’il contient plusieurs lignes. 10relief
Spécifie le type de bordure. Sa valeur par défaut est FLAT.
10. `selectbackground` : la couleur d’arrière-plan du texte sélectionné.
11. `selectborderwidth` : définit la largeur de la bordure à utiliser autour du texte sélectionné. La valeur par défaut est un pixel.
12. `selectforeground` : définit la couleur de premier plan du texte sélectionné.
13. `show `: est utilisé pour afficher ou masquer les caractères. Exemple pour un mot de
passe on utilise show = "*".
14. `state` : la valeur par défaut est state = NORMAL, mais vous pouvez utiliser state =
DISABLED pour masquer le contrôle et le rendre inactif.
15. `textvariable` : Pour pouvoir extraire le texte actuel de votre widget Entry, vous devez
définir cette option sur une instance de la classe StringVar.
16. `width` : définit la largeur du widget
17. `xscrollcommand` : ajoute une barre de défilement au widget.

### **3.3.2 Entry (avec StringVar)**

**3️⃣ set() – Définir une valeur**

La méthode **set() **est utilisée avec :

- StringVar()

- IntVar()

- DoubleVar()

- BooleanVar()

- ttk.Combobox

**📌 Exemple avec StringVar**

````python

import tkinter as tk

fenetre = tk.Tk()

# entrée
value = tk.StringVar()
entree = tk.Entry(fenetre, textvariable=value, width=30)
entree.pack()


value.set("texte par défaut")

def afficher_texte():
    print(value.get())

btn = tk.Button(fenetre, text="Afficher", command=afficher_texte)
btn.pack()


fenetre.mainloop()
````

👉 Ici on ne fait pas entry.insert(), on modifie directement la variable liée.

### **3.3.3 configure() / config()**

### **3.3.5 Frame**

En Tkinter, **Frame** et **LabelFrame** sont des conteneurs permettant d’organiser les widgets (boutons, labels, entrées, etc.).

**🔹 1) Frame dans Tkinter**

**Frame** est un conteneur simple utilisé pour regrouper et organiser des widgets.

````python
import tkinter as tk

root = tk.Tk()
root.title("Exemple Frame")

# Création d'un Frame
frame = tk.Frame(root)
frame.pack()

# Widgets dans le Frame
label = tk.Label(frame, text="Bonjour Tkinter")
label.pack(pady=5)

button = tk.Button(frame, text="Cliquez ici")
button.pack(pady=5)

root.mainloop()
````

**✅ Explication :**

**- bg** → couleur de fond

**- bd**→ épaisseur de bordure

**- relief** → style de bordure (solid, ridge, groove, etc.)

**- pack()** → méthode de placement

**🔹 2) LabelFrame dans Tkinter**

**LabelFrame** est comme un Frame, mais avec un titre intégré.

Il est très utile pour regrouper des champs (ex: formulaire).

````python
import tkinter as tk

root = tk.Tk()
root.title("Exemple LabelFrame")

# Création d'un LabelFrame
labelframe = tk.LabelFrame(root, text="Informations personnelles",
                           #labelanchor="ne",
                            padx=10, pady=10)
#labelframe.pack(padx=20, pady=20)
labelframe.pack(expand=True, fill="both", padx=20, pady=20)

# Widgets dans le LabelFrame
tk.Label(labelframe, text="Nom:").grid(row=0, column=0)
tk.Entry(labelframe).grid(row=0, column=1)

tk.Label(labelframe, text="Prénom:").grid(row=1, column=0)
tk.Entry(labelframe).grid(row=1, column=1)

root.mainloop()
````

**✅ Explication :**

**- text="..."** → titre affiché sur la bordure

**- padx / pady** → espace intérieur

**- grid()** → disposition en tableau

**- fill** → Permet au widget de s’étirer dans une direction donnée.

**- labelanchor**  → dans LabelFrame permet de positionner son titre.

**- anchor** → dans Label permet de positionner  son texte

**- anchor** → dans pack permet de positionner son widget 

Valeurs possibles :

   - fill="x" → s’étire horizontalement

   - fill="y" → s’étire verticalement

   - fill="both" → s’étire horizontalement et verticalement

   - fill=None → pas d’étirement

**- expand** →  Permet au widget de prendre l’espace supplémentaire lorsque la fenêtre est agrandie.

   - expand=False (par défaut) → ne prend pas l’espace libre

   - expand=True → occupe l’espace libre disponible

**🎯 Exemple combiné (Frame + LabelFrame)**

````python
import tkinter as tk

root = tk.Tk()
root.title("Exemple combiné")

main_frame = tk.Frame(root)
main_frame.pack(padx=20, pady=20)

info_frame = tk.LabelFrame(main_frame, text="Connexion")
info_frame.pack(pady=10)

tk.Label(info_frame, text="Email:").grid(row=0, column=0)
tk.Entry(info_frame).grid(row=0, column=1)

tk.Label(info_frame, text="Mot de passe:").grid(row=1, column=0)
tk.Entry(info_frame, show="*").grid(row=1, column=1)

root.mainloop()
````

### **3.3.6 Toplevel**

En Tkinter, **une fenêtre Toplevel** est une fenêtre secondaire créée à partir de la fenêtre principale de l’application. 

Elle sert souvent pour les boîtes de dialogue, les fenêtres de paramètres, ou les fenêtres popup.

**Exemple 1**

````python
import tkinter as tk

root = tk.Tk()
root.title("Fenêtre principale")
root.geometry("300x200")

top1 = tk.Toplevel(root)   # création de la fenêtre secondaire
top1.title("Fenêtre secondaire")
top1.geometry("200x150")

top2 = tk.Toplevel(root)   # création de la fenêtre tertiaire
top2.title("Fenêtre tertiaire")
top2.geometry("200x150")

root.mainloop()
````

## **3.10 Menu (barre de menus)**

Ajoute un **menu en haut** (Fichier, Aide, etc.).


````python
import tkinter as tk
from tkinter import messagebox

fenetre = tk.Tk()
fenetre.title("Menu")


def nouveau():
    messagebox.showinfo("Nouveau", "Nouveau document")


def quitter():
    fenetre.destroy()


def apropos():
    messagebox.showinfo("À propos", "Exemple de menu Tkinter")


# Barre de menus
menubar = tk.Menu(fenetre)

menu_fichier = tk.Menu(menubar, tearoff=0)
menu_fichier.add_command(label="Nouveau", command=nouveau)
menu_fichier.add_separator()
menu_fichier.add_command(label="Quitter", command=quitter)

menu_aide = tk.Menu(menubar, tearoff=0)
menu_aide.add_command(label="À propos", command=apropos)

menubar.add_cascade(label="Fichier", menu=menu_fichier)
menubar.add_cascade(label="Aide", menu=menu_aide)

fenetre.config(menu=menubar)

# Contenu
label = tk.Label(fenetre, text="Regardez la barre de menus en haut 👆")
label.pack(padx=20, pady=30)

fenetre.mainloop()
````

## bind()

**👉 Sert à associer un événement à une fonction.**

Contrairement à command= (qui fonctionne surtout avec Button), bind() permet de gérer tous les événements clavier et souris.

````python
import tkinter as tk

fenetre = tk.Tk()


def clic_souris(event):
    print("Position :", event.x, event.y)


fenetre.bind("<Button-1>", clic_souris) # clic gauche sur la fenêtre

fenetre.mainloop()
````

````python
import tkinter as tk

def clic(event):
    print("Coordonnées :", event.x, event.y)

root = tk.Tk()
label = tk.Label(root, text="Clique ici", bg="lightblue")
label.pack(padx=20, pady=20)

label.bind("<Button-1>", clic) # clic gauche sur le label

root.mainloop()
````

# 📌 8. Application Login avec Tkinter

Fonctionnalités demandées :


- Label et champ de saisie Username

- Label et champ de saisie Password

- Boutons :

    - Login

    - New User

    - Forgot Password

- Stockage des utilisateurs dans un fichier texte **users.txt :  username;password**

````python
import tkinter as tk
from tkinter import messagebox

FILE = "users.txt"

# ------------------ Fonctions ------------------

def new_user():
    username = entry_user.get()
    password = entry_pass.get()

    if username == "" or password == "":
        messagebox.showwarning("Erreur", "Champs vides")
        return

    with open(FILE, "a") as f:
        f.write(username + ";" + password + "\n")

    messagebox.showinfo("Succès", "Utilisateur ajouté")


def login():
    username = entry_user.get()
    password = entry_pass.get()

    try:
        with open(FILE, "r") as f:
            users = f.readlines()
    except FileNotFoundError:
        messagebox.showerror("Erreur", "Fichier introuvable")
        return

    for user in users:
        u, p = user.strip().split(";") # on enlève le \n et on sépare username et password
        if u == username and p == password:
            messagebox.showinfo("Login", "Connexion réussie")
            return

    messagebox.showerror("Login", "Identifiants incorrects")


def forgot_password():
    username = entry_user.get()

    try:
        with open(FILE, "r") as f:
            users = f.readlines()
    except FileNotFoundError:
        messagebox.showerror("Erreur", "Fichier introuvable")
        return

    for user in users:
        u, p = user.strip().split(";")
        if u == username:
            messagebox.showinfo("Mot de passe", "Votre mot de passe est : " + p)
            return

    messagebox.showerror("Erreur", "Utilisateur non trouvé")


# ------------------ Interface ------------------

root = tk.Tk()
root.title("Login System")
root.geometry("300x250")

tk.Label(root, text="Username").pack(pady=5)
entry_user = tk.Entry(root)
entry_user.pack()

tk.Label(root, text="Password").pack(pady=5)
entry_pass = tk.Entry(root, show="*")
entry_pass.pack()

tk.Button(root, text="Login", command=login).pack(pady=5)
tk.Button(root, text="New User", command=new_user).pack(pady=5)
tk.Button(root, text="Forgot Password", command=forgot_password).pack(pady=5)

root.mainloop()
````
