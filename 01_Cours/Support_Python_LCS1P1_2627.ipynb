#### **3.2. Les boucles :**
##### ***3.2.1. La boucle while*** :
L’instruction while (“tant que”, en français) permet d’exécuter une boucle tant qu’une condition est vraie.

**Exemple 2:**

````python
i = 1 # i vaut 1 la valeur initiale(VI)=1
print("valeur de i avant la boucle : " , i)
while i <= 10: #ici la valeur finale (VF)=10
  print("valeur de i dans la boucle : " , i)
  i = i + 1 # on incrémente i d’un pas de 1
print("valeur de i après la boucle : " , i)
````

**Exemple 3:**

````python
taux = 0.03
capital = 1000.0
annee = 2016
while annee < 2020:
    annee = annee + 1
    capital = capital * (1 + taux)
print(annee, capital)
````

#### ***3.3. Les instructions break et continue :***
Il est possible de sortir d’une boucle avec l’instruction break. Cette instruction est très pratique pour tester une condition d’arrêt qui dépend d’une valeur entrée.

**Exemple 5:**

````python
somme = 0
while True:
   n = int(input('Entrez un nombre (0 pour arrêter): '))

   if n == 0:

      break
   somme = somme + n
print("La somme des nombres est", somme)
````

La fonction continue : retourne directement au début de la boucle, ce qui permet d'éviter des tests de condition inutiles.

**Exemple 6:**

````python
nb=9
while nb!=0:
  nb=int(input("Entrer 0 , 1, 2 ou 3? : "))
  if nb==1:
    print ("Choix un");
    continue
  if nb==2:
    print ("Choix deux");
    continue
  if nb==3:
   print ("Choix trois")
````

### **4. Les fonctions en python**
Une fonction est un sous programme qui réalise une certaine tâche.
Une fonction est composée de trois grandes parties :
 1. Son nom qui permet d'y faire appel et l’identifier des autres fonctions.
 2. Ses arguments qui permettent de spécifier des données à lui transmettre.
 3. Sa sortie, c'est-à-dire ce qu'elle retourne comme résultat.

 Son nom et ses arguments forment ce que l'on appelle la **signature** d'une fonction.
 Deux fonctions seront différenciées par l'interpréteur à partir du moment où elles ne possèdent pas la même signature.

 * ***Les fonctions*** qui effectuent des instructions et retournent un résultat.
 * ***Les procédures*** qui effectuent des instructions mais ne retournent rien du tout.

**Exemple 7:**

````python
def mafonction():
    print(' ma fonction ')
# pp
mafonction()
````

````python
def mafonction(a,b):
    print(f"Ma  fonction {a} puissance {b} est {a**b}")
# pp
a=2
b=3
mafonction(a,b)
````

````python
def mafonction(a,b):
       return a**b
# pp
a=2
b=4
t=mafonction(a,b)
print(f"{a} puissance {b} = {t} ")
````

Les fonctions **lambda** sont des fonctions qui ne sont pas définies avec un nom, mais qui peuvent être utilisées dans des expressions ou des fonctions.

Les fonctions lambda avec Python sont créées selon la syntaxe suivante :

**lambda** arguments : expression

````python
f=lambda x,y:x**y
x=2
y=3
f(x,y)
````

Les fonctions avec des arguments ayant des valeurs par défaut

````python
def mafonction(a,b=3):
       return a**b
# pp
a=2
# Si la valeur du 2ème paramètre est manquée la fonction prend
# la valeur par défaut qui est b=3
mafonction(a)
print(f"{a} puissance {b} = {mafonction(a,b)} ")
````

#### **4.1. Les fonctions avec un nombre d’arguments indéfini**
##### **4.1.1. Arguments anonymes**

**Exemple 8:**

````python
def fonction(*arguments):
    for element in arguments:
        print(element)
# pp
fonction('Ali', 20, 'LCS1')
````

````python
def fonction(*arguments):
    for element in arguments:
        print(element)
# pp
# fonction('Ali 20 LCS1')
fonction(*'Ali 20 LCS1')
````

##### **4.1.2. Arguments avec clé :**
Si on veut associer une clé à chacun des arguments qu’on donne en supplément à une fonction, il va falloir utiliser un dictionnaire (voir Exemple ). Et ceci, c'est l'opérateur '**' qui sera utilisé.

**Exemple 9:**

````python
def fonction(**arguments):
    for cle in arguments:
        print(cle,arguments[cle])
# pp
fonction(arg1='Ali',arg2=20, arg3= 'LCS1')
````

Une autre façon d’appeler la fonction : remarquez la présence du double étoiles ** et le mot **dict**.

````python
def fonction(**arguments):
    for cle in arguments:
        print(cle,arguments[cle])
# pp
fonction(**dict(arg1='Ali',arg2=20, arg3= 'LCS1'))
````

##### **4.1.3. Documenter une fonction :**
Une dernière chose très importante est la documentation de la fonction. On va commencer à documenter notre code pour le rendre plus compréhensible et exploitable par d'autres personnes.
Pour cela, c'est très simple, il suffit de placer immédiatement sous la ligne de définition de notre fonction, une chaîne de caractères entre triple 'quote' (""" ou ''') qui sera considérée un commentaire.

**Exemple 10:**

````python
def fonction(arg1, arg2, arg3):
    ''' j'ai documente ma fonction
    arg1: pour ...
    arg2: pour ...
    arg3: pour ...'''
    pass
# pp
fonction(1,2,3)
````

### **5.2.6. Les méthodes d’une liste**
Une fois définie, une liste possède un ensemble de méthodes :

***append*** : Pour ajouter un élément à la fin d’une liste.

````python
L=[2, 4, 6, 8, 10, 12, 14]
L.append(16)
L      # affichage [2, 4, 6, 8, 10, 12, 14, 16]
````

***clear :*** Pour supprimer tous les éléments de la liste = vider la liste.

````python
L=[2, 4, 6, 8, 10, 12, 14]
L.clear()
L      # affichage []
````

***Copy :*** pour copier une liste dans une autre

````python
L=[2, 4, 6, 8, 10, 12, 14]
L1=L
print(L1)
print(L)
````

````python
L=[2, 4, 6, 8, 10, 12, 14]
L1=L
L1[0]=50
print(L1)
print(L)

# affichage changement d'un élément de L1
# L1= [50, 4, 6, 8, 10, 12, 14]
# L=[50, 4, 6, 8, 10, 12, 14]
# Les deux objets pointent vers une même adresse
# le changement de l’un des objets affecte l’autre.
````

````python
L=[2, 4, 6, 8, 10, 12, 14]
L1=L
L[6]=100
print(L1)
print(L)

# affichage changement d'un élément de L
# L1= [2, 4, 6, 8, 10, 12, 100]
# L= [2, 4, 6, 8, 10, 12, 100]
# Les deux objets pointent vers une même adresse
# le changement de l’un des objets affecte l’autre.
````

Demandez à l’utilisateur de saisir deux séries de nombres.

Puis :

* Transformer les deux séries en deux ensembles

* Affichez les nombres qui sont en commun dans les deux séries

* Affichez le plus grand nombre qui n’est présent que dans une seule série

* Affichez tous les nombres sans doublon.

````python
# Saisir les deux séries de nombres
serie1 = list(map(int, input("Saisir la première série de nombres (séparés par des espaces) : ").split()))
serie2 = list(map(int, input("Saisir la deuxième série de nombres (séparés par des espaces) : ").split()))

# 1. Transformer en ensembles pour faciliter le traitement
set1 = set(serie1)
set2 = set(serie2)

# 2. Nombres en commun
communs = set1 & set2
print("Nombres en commun :", communs)

# 3. Plus grand nombre présent dans une seule série
seulement_une = (set1 ^ set2)  # éléments présents dans une seule série
if seulement_une:
    print("Plus grand nombre présent dans une seule série :", max(seulement_une))
else:
    print("Il n'y a aucun nombre présent dans une seule série")

# 4. Tous les nombres sans doublon
sans_doublon = set1 | set2
print("Tous les nombres sans doublon :", sans_doublon)
````

La méthode **d.setdefault(k[ x])** en Python est utilisée pour obtenir la valeur d'une clé donnée dans un dictionnaire. Si la clé n'existe pas, elle est ajoutée avec une valeur par défaut.

**1. La clé existe déjà**

````python
d = {'a': 1, 'b': 2}
val = d.setdefault('a', 100)

print(val)  # Output: 1
print(d)    # Output: {'a': 1, 'b': 2}
# La clé 'a' existe, donc sa valeur 1
# est retournée et d reste inchangé.
````

**2. La clé n'existe pas, ajout avec valeur par défaut :**

````python
d = {'x': 10, 'y': 20}
val = d.setdefault('z', 30)

print(val)  # Output: 30
print(d)    # Output: {'x': 10, 'y': 20, 'z': 30}
# La clé 'z' n'existait pas, donc elle est ajoutée
# avec la valeur 30.
````

**3. La clé n'existe pas et aucune valeur par défaut n'est fournie :**

````python
d = {'p': 5, 'q': 6}
val = d.setdefault('r')

print(val)  # Output: None
print(d)    # Output: {'p': 5, 'q': 6, 'r': None}
 # Comme 'r' n'existait pas et qu'aucune valeur
 # par défaut n'était spécifiée, None est utilisé.
````

**2. Suppression successive des éléments :**

````python
d = {'x': 10, 'y': 20}

print(d.popitem())  # Output: ('y', 20)
print(d.popitem())  # Output: ('x', 10)
print(d)            # Output: {}
````

d1={clé1 :val1, clé2 :val2, clé3 :val3, …….}

````python
d = {'nom': 'Alice', 'âge': 25, 'ville': 'Paris'}
print(d)  # Output: {'nom': 'Alice', 'âge': 25, 'ville': 'Paris'}
````

````python
cles = ['nom', 'âge', 'ville']
valeurs = ['Alice', 25, 'Paris']
d = dict(zip(cles, valeurs))

print(d)  # Output: {'nom': 'Alice', 'âge': 25, 'ville': 'Paris'}
````

**1. Saisir les elements de l'examen avec input**

````python

Examen={}
Examen["code"] = input("Saisir le code de l'examen : ")
Examen["libelle"] = input("Saisir le libelle de l'examen : ")

Examen["date"] = {}
Examen["date"]["jour"] = int(input("Saisir le jour de l'examen : "))
Examen["date"]["mois"] = int(input("Saisir le mois de l'examen : "))
Examen["date"]["annee"] = int(input("Saisir l'annee de l'examen : "))

Examen["module"] = {}
Examen['module']['code'] = input("Saisir le code du module : ")
Examen['module']['nom'] = input("Saisir le nom du module : ")
print(Examen)
````

**2. Modifier une valeur imbriquée**

````python
Examen["date"]["annee"] = 2027
print(Examen)
````

**3. Saisir plusieurs examens dans une liste Examens**

````python
def saisir_examens(n):
    Examens = []

    for i in range(n):
        print(f"\n--- Saisie de l'examen {i+1} ---")

        Examen = {
            "code": input("Code examen : "),
            "libelle": input("Libellé : "),
            "date": {
                "jour": int(input("Jour : ")),
                "mois": int(input("Mois : ")),
                "annee": int(input("Année : "))
            },
            "module": {
                "code": input("Code module : "),
                "nom": input("Nom module : ")
            }
        }

        Examens.append(Examen)

    return Examens
````

````python
n = int(input("Donner le nombre d'examens : "))
Examens = saisir_examens(n)

print(Examens)
````

**Version avec enumerate pour l’affichage**

````python
for i, examen in enumerate(Examens, start=1):
    print(f"\nExamen {i}")
    print("Code :", Examen["code"])
    print("Libellé :", Examen["libelle"])
    print("Date :", Examen["date"]["jour"], "/",
                    Examen["date"]["mois"], "/",
                    Examen["date"]["annee"])
    print("Module :", Examen["module"]["nom"])
````

### **5.6.3. les méthodes d’une chaine de caractères**

***Split()*** : découpe une chaine en une liste de mots

````python
S= "Bonjour tout le monde"
LS=S.split()
print(LS)   #  Output:  ['Bonjour', 'tout', 'le', 'monde']
````

***Join(liste de chaines)*** : concatène une liste de chaine en chaine unique

````python
S1="-"
S="Bonjour Tout Le Monde"
S3=S1.join(S)
print(S3)    #  Output:  B-o-n-j-o-u-r- -T-o-u-t- -L-e- -M-o-n-d-e
````

````python
S1="-"
S4=S1.join(['Bonjour','tout','le','monde'])
print(S4)    #  Output:  Bonjour-tout-le-monde
````

````python
S5= " ".join(['Bonjour','tout','le','monde'])
print(S5)   #  Output:  Bonjour tout le monde
````

***Find(sous chaine)*** : donne la position d’une sous chaine dans une chaine

````python
S="Bonjour Tout Le Monde"
f=S.find('tout')
f  # Output: -1 la sous chaine n’existe pas
````

````python
S="Bonjour Tout Le Monde"
f=S.find('Tout')
f  # Output: 8 la sous chaine existe
````

***Count(sous chaine)*** : donne le nombre le nombre d’apparition d’une sous chaine dans une chaine

````python
S="Bonjour Tout Le Monde"
f=S.count('tout')
f  # Output: 0 la sous chaine n’existe pas
````

````python
S="Bonjour Tout Le Monde"
f=S.count('Tout')
f    # Output: 1 la sous chaine existe
````

***lower()*** : convertit une chaine en minuscule

````python
S="Bonjour Tout Le Monde"
Sl=S.lower()
print(Sl)    # Output:  bonjour tout le monde
````

***upper()*** : convertit une chaine en majuscule

````python
S="Bonjour Tout Le Monde"
Su=S.upper()
print(Su)    # Output:  BONJOUR TOUT LE MONDE
````

***capitalize()*** : convertit la 1ere lettre d’une chaine en majuscule.

````python
S="Bonjour Tout Le Monde"
SM=S.capitalize()
print(SM)  # Output: Bonjour tout le monde
````

***title()*** : convertit tous les 1er lettres des mots d’une chaine en majuscule.

````python
S="bonjour tout le monde"
St=S.title()
print(St)  # Output: Bonjour Tout Le Monde
````

***swapcase()*** : intervertit les lettres majuscules et minuscules

````python
S="Bonjour Tout Le Monde"
Si=S.swapcase()
print(Si)  # Output: bONJOUR tOUT lE mONDE
````

***strip()***: supprime les espaces blancs en début en en fin de la chaine

````python
S="  bonjour tout le monde   "
Ss=S.strip()
print(Ss)  # Output: bonjour tout le monde
````

***replace()*** : remplace une sous chaine par une autre

````python
S="Bonjour Tout Le Monde"
S1=S.replace('Bonjour','Bonsoir')
print(S1) # Output: Bonsoir Tout Le Monde
````

***index(sous chaine)*** : retourne la position de la sous chaine dans une chaine

````python
S="Bonjour"
S.index('j') # Output: 3
````

````python
S="Bonjour"
S.index('njour') # Output: 2
````

***center(n)*** : centrer la chaine sur un nombre n caractères

````python
S="Bonjour"
print(S.center(40))  # Output:             Bonjour
````

***format()*** : remplace un format dans une chaine

````python
S1="Voici {0} chaîne à {1} trous."
print(S1)  # Output:   Voici {0} chaîne à {1} trous.
````

````python
S1="Voici {0} chaîne à {1} trous."
print(S1.format("une " , 2)) # Output:  Voici une chaîne à 2 trous.
````

````python
print("a{0}cada{0}".format("bra"))
# Output: abracadabra
````

### **7. Gestion des erreurs avec `try / except`**

Un programme doit prévoir les entrées incorrectes et certaines erreurs d’exécution.

```python
try:
    age = int(input("Entrez votre âge : "))
    print(f"Vous avez {age} ans.")
except ValueError:
    print("Erreur : veuillez saisir un entier.")
```

On peut ajouter `else` et `finally`.

````python
try:
    note = float(input("Entrez une note sur 20 : "))
    if not 0 <= note <= 20:
        raise ValueError("La note doit être comprise entre 0 et 20.")
except ValueError as erreur:
    print("Erreur :", erreur)
else:
    print(f"Note enregistrée : {note}/20")
````
