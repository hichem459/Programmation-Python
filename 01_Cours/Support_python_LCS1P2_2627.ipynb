## **Numpy**

Le Python pur est peu performant pour le calcul. Les listes ne sont pas des objets efficaces pour représenter les tableaux numériques de grande taille. [Numpy](http://www.numpy.org/) a été créé à l'initiative de développeurs qui souhaitaient combiner la souplesse du langage python et des outils de calcul algébrique performants.

Numpy se base sur :

- le `ndarray` : tableau multidimensionnel
- des objets dérivés comme les *masked arrays* et les matrices
- les `ufunc`s : opérations mathématiques optimisées pour les tableaux
- des méthodes pour réaliser des opération rapides sur les tableaux :
    - manipulation des *shapes*
    - tri
    - entrées/sorties
    - FFT
    - algébre linéaire
    - statistiques
    - calcul aléatoire
    - et bien plus !
  
Numpy permet de calculer *à la Matlab* en Python. Il est un élément de base de l'écosystème [SciPy](https://www.scipy.org/)

- un `ndarray` a une taille fixée à la création
- un `ndarray` est composé d'éléments du même type

## Opérations basiques sur les tableaux

Par défaut, Numpy réalise les opérations arithmétiques éléments par éléments

````python
a = np.array([0, 1, 2, 3])
b = np.array((4, 5, 6, 7))

print(f'a*b = {a*b}')  # Produit éléments par éléments : pas le produit matriciel !
print(f'a + b = {a + b}')  # Addition élément par élément : pas l'addition matricielle !
````

````python
print(a**2) # Élève chaque élément au carré
````

````python
print(5*a) # Multiplie chaque élément par 5
print(5 + a) # Ajoute 5 à chaque élément
````

````python
print(a < 2) # Retourne un tableau de booléens : True si l'élément est inférieur à 2, False sinon
````

````python
print(np.cos(a*np.pi))  # Utilisation de ufunc
````

**🔢 ufunc en Python (NumPy)**

Une ufunc (Universal Function) est une fonction de NumPy qui applique une opération élément par élément sur des tableaux (ndarray).

👉 Elle est rapide, vectorisée et fonctionne sur des tableaux de n’importe quelle taille.

````python

a = np.array([1, 2, 3, 4])

print(np.sqrt(a))     # racine carrée
print(np.exp(a))      # exponentielle
print(np.add(a, 10))  # addition élément par élément
````

##  Indexation et slicing

### Indexation

Les règles différent légèrement des listes pour les tableaux multi-dimensionnels

````python
# Indexation d'un tableau numpy
a = np.arange(4).reshape(2, 2)  # Notez l'effet de la méthode reshape()
print(a)
b=np.arange(9)
print(b[0])
````

````python
# Indexation de la liste équivalente
liste = a.tolist()
print(liste)
print(liste[1][1])
````

### Slicing

Pour les tableaux unidimensionnels, les règles de slicing sont les mêmes que pour les listes.
Pour les tableaux multidimensionnels numpy, le slicing permet d'extraire des séquences dans n'importe quelle direction.

````python
print(a)
print(a[2, :])  # Retourne la 3ème ligne
print(a[:, 1])  # Retourne la deuxième colonne
````

**Attention :** contrairement aux listes, le slicing de tableaux ne renvoie pas une copie mais constitue une ***vue*** du tableau.

````python
b = a[:, 1]
b[0] = 100
print(a)
````

`a` est aussi une vue du tableau `np.arange(9)` obtenue avec la méthode `reshape()` donc `a` et `b` sont deux vues différentes du même tableau :

````python
b.base is a.base  # b.base retourne le tableau dont b est la vue
````

Si on veut réaliser une copie d'un tableau, il faut utiliser la fonction `copy()`

````python
b = a[:, 1].copy()
````

ici `b` n'est pas une vue mais une copie donc `b.base` retourne `None` et `a` n'est pas modifié.

````python
print(b.base)
b[0] = 200
print(a)
````

## Quelques opérations d'algèbre linéaire

````python
A = np.array([[1, 1],[0, 1]])
B = np.array([[2, 0],[3, 4]])

print(A.dot(B))      # produit matriciel
print(np.dot(A, B))  # une autre syntaxe de produit matriciel
print(A@B)           # encore un autre produit matriciel
````

````python
a = np.array([[1.0, 2.0],
              [3.0, 4.0]])
print(a)
````

````python
# Deux syntaxes équivalentes pour la transposition
print(a.transpose())
print(a.T)
````

````python
print(np.linalg.inv(a))  # inversion de matrice
````

````python
print(np.trace(a))  # trace
````

````python
print(np.eye(2))     # "eye" représente "I", la matrice identité
````

````python
y = np.array([[5.], [7.]])  # Résoudre A*x = y
x = np.linalg.solve(a, y)
print(np.linalg.solve(a, y))
print(a@x == y)
````

````python
j = np.array([[0.0, -1.0], [1.0, 0.0]])
print(np.dot(j, j))     # produit matriciel
print(np.linalg.eig(j))  # Extraction des valeurs propres
````

**affichage de plusieurs courbes avec subplot en Python**

On peut utiliser Matplotlib pour afficher plusieurs courbes dans des sous-graphes avec plt.subplot().

````python
import numpy as np
import matplotlib.pyplot as plt

# Création des données
x = np.linspace(-10, 10, 100)
y1 = x**2  # Parabole
y2 = np.sin(x)  # Sinusoïde

# Création des subplots (2 lignes, 1 colonne)
plt.figure(figsize=(8, 6))

# Premier subplot : y = x^2
plt.subplot(2, 1, 1)  # 2 lignes, 1 colonne, 1er subplot
plt.plot(x, y1, 'r*', label="y = x²")
plt.title("Parabole")
plt.legend()
plt.grid()

# Deuxième subplot : y = sin(x)
plt.subplot(2, 1, 2)  # 2 lignes, 1 colonne, 2e subplot
plt.plot(x, y2, 'bo', label="y = sin(x)")
plt.title("Sinusoïde")
plt.legend()
plt.grid()

# Affichage des graphes
plt.tight_layout()
plt.show()
````

# **Les Fichiers**

Une grande partie de l’information est stockée sous forme de texte dans des fichiers. Pour traiter cette information, vous devez le plus souvent lire ou écrire dans un ou plusieurs fichiers. Python possède pour cela de nombreux outils qui vous simplifient la vie.

**Méthode .readlines()**

Avant de passer à un exemple concret, créez un fichier dans l’éditeur de texte de votre choix avec le contenu suivant :

girafe

tigre

singe

souris

Enregistrez ce fichier dans votre répertoire courant avec le nom a**nimaux.txt**.

````python
file = open("animaux.txt", "r")
file.readlines() 
file.close()

file.readlines()  # il ne peut pas lire un fichier fermé
````

Voici maintenant un exemple complet de lecture d’un fichier avec Python :

````python
file = open("animaux.txt", "r")
lignes = file.readlines()
for ligne in lignes:
   print(ligne)
file.close()
````

**Méthode .read()**

Il existe d’autres méthodes que **.readlines()** pour lire (et manipuler) un fichier.
 
Par exemple, la méthode .read() lit tout le contenu d’un fichier et renvoie une chaîne de caractères unique :

````python
with open("animaux.txt", "r") as file:
     r=file.read()
     print(r)
````

**Méthode .readline()**

La méthode **.readline()** (sans s à la fin) lit une ligne d’un fichier et la renvoie sous forme de chaîne de caractères.

À chaque nouvel appel de .readline(), la ligne suivante est renvoyée. Associée à la boucle while, cette méthode permet de lire un fichier ligne par ligne :

````python
with open("animaux.txt", "r") as file:
    ligne = file.readline()
    while ligne != "":
        print(ligne)
        ligne = file.readline()
````

**Écriture dans un fichier**

Écrire dans un fichier est aussi simple que de le lire. Voyez l’exemple suivant :

````python
animaux2 = ["poisson  ", "abeille  ", "chat "]
with open("animaux2.txt", "w") as file:
      for animal in animaux2:
        # file.write(animal)
         file.write(f"{animal}\n")  # saut de ligne
````
